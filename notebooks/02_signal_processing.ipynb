# 02 — Signal processing

FFT of a simulated bearing-degradation window.

```python
from machine_failure_investigator.machine.faults import FaultSpec, FaultType
from machine_failure_investigator.machine.model import RotatingMachine
from machine_failure_investigator.simulation.simulator import MachineSimulator
from machine_failure_investigator.signal_processing.fft import compute_fft

bundle = MachineSimulator().simulate(
    RotatingMachine().state(fault=FaultSpec(FaultType.BEARING_DEGRADATION, 0.6))
)
freqs, mag = compute_fft(bundle.vibration, bundle.sample_rate)
print(freqs[mag.argmax()], mag.max())
```
