# 04 — Anomaly detection

```python
import numpy as np
from machine_failure_investigator.models.anomaly.isolation_forest import IsolationForestDetector

rng = np.random.default_rng(0)
X_h = rng.normal(size=(100, 8))
X_a = rng.normal(loc=3.0, size=(20, 8))
det = IsolationForestDetector(n_estimators=50, random_state=0).fit(X_h)
print(det.predict(X_a).mean())
```
